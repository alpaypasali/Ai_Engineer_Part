```python
x = torch.tensor(2.0, requires_grad=True)

# İlk hesaplama
y = x**2
y.backward()

print("1. Gradient:", x.grad)

# İkinci hesaplama
z = 3*x**2
x.grad.zero_()
z.backward()

print("2. Gradient:", x.grad)
```

```python
import torch

w = torch.tensor(5.0, requires_grad=True)

optimizer = torch.optim.SGD([w], lr=0.1)

loss = w ** 2

optimizer.zero_grad()

loss.backward()

print("Gradient:", w.grad.item())  # 10.0
print("Backward sonrası ağırlık:", w.item())  # 5.0

optimizer.step()

print("Yeni ağırlık:", w.item())  # 4.0
```

```python
x = torch.tensor(
    2.0,
    requires_grad=True
)

y = x**2

z = y.detach()

result = z * 3

result.backward()

print(x.grad)
```
