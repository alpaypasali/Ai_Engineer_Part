```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import re
from collections import Counter
```

```python
df = pd.read_csv("../datas/IMDB Dataset.csv")
df.head()
```

```python
documents = df["review"]
labels = df["sentiment"]
```

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(r"\d+" , "" , text)

    text = re.sub(r"[^\w\s]" , "" , text)

    text = " ".join([word for word in text.split() if len(word) > 2])

    return text
```

```python
cleaned_doc = [clean_text(doc) for doc in documents]
```

```python
vectorizer = CountVectorizer()
```

```python
X = vectorizer.fit_transform(cleaned_doc[:100])
```

```python
feature_names = vectorizer.get_feature_names_out()
```

```python
feature_names
```

```python
print(X.toarray())
```

```python
df_bag_of_words = pd.DataFrame(X.toarray() , columns=feature_names)
```

```python
df_bag_of_words.head(20)
```

```python
word_count = X.sum(axis=0).A1
```

```python
word_freq = dict(zip(feature_names , word_count))
```

```python
word_freq
```

```python
most_common_5_words = Counter(word_freq).most_common(5)
```
